```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Génération de la série temporelle
## Temps

```python
n = 50  # Nombre de points
t = np.arange(n)
```

## Tendance affine 

```python
a = 0.5  
b = 10   
trend = a * t + b
```

## Saisonnalité de période 4 (sinusoïdale simplifiée)

```python
seasonality = 2 * (np.sin(2 * np.pi *t /4))
```

## Bruit blanc (moyenne 0, variance 1)

```python
np.random.seed(42)
noise = np.random.normal(0,1,n)
```

## Série résultante

```python
series = trend + seasonality + noise
```

# Estimation de la tendance par moyenne mobile d'ordre 4

```python
rolling_mean = pd.Series(series).rolling(window=4, center=True).mean()
```

# Tracé des résultats

```python
plt.figure(figsize=(10, 5))
plt.plot(t, series, label="Série simulée (brute)", color="black", alpha=0.6)
plt.plot(t, rolling_mean, label="Tendance lissée", color="red", linewidth=2)
plt.legend()
plt.xlabel("Temps (t)")
plt.ylabel("Valeur")
plt.title("Simulation et lissage d'une série temporelle")
plt.show()
```
